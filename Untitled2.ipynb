```python
import pandas as pd
import numpy as np
```

```python
# 1
df = pd.read_csv("waste_collection_records.csv")
```

```python
print("Original dataset:")
display(df.head())
```

```python



# Step 1 — Standardize dates (YYYY-MM-DD)
dates = df['collection_date'].astype(str).str.strip().replace({'nan': None, 'NaN': None, '': None})

def parse_mixed_date_debug(date_str):
    if pd.isna(date_str):
        return pd.NaT
    if '-' in date_str:
        sep = '-'
    elif '/' in date_str:
        sep = '/'
    else:
        return pd.NaT
    parts = date_str.split(sep)
    if len(parts) != 3:
        return pd.NaT
    first = parts[0]
    if len(first) == 4 and first.isdigit():
        fmt = f'%Y{sep}%m{sep}%d'
    elif len(first) <= 2 and first.isdigit():
        fmt = f'%d{sep}%m{sep}%Y'
    else:
        return pd.NaT
    return pd.to_datetime(date_str, format=fmt, errors='coerce')

df['collection_date'] = dates.apply(parse_mixed_date_debug)
df['collection_date'] = df['collection_date'].dt.date

print("\n Date formats standardized to YYYY-MM-DD")

df.head()
```

```python
# Step 2 — Remove duplicates
df = df.drop_duplicates()
```

```python
# step 3
df['client_name'] = df['client_name'].fillna('Unknown Client')

df.head()
```

```python
# Step 4
def convert_to_kg(value):
    if pd.isna(value):
        return None
    value = str(value).strip().lower()
    try:
        if "ton" in value:
            num = float(value.replace("tons", "").replace("ton", "").strip())
            return num * 1000
        elif "lb" in value:
            num = float(value.replace("lbs", "").replace("lb", "").strip())
            return num * 0.4536
        elif "kg" in value:
            num = float(value.replace("kg", "").strip())
            return num
        else:
            return float(value)
    except:
        return None


weight_col = [c for c in df.columns if 'weight' in c][0]
print(f"\nDetected weight column: {weight_col}")

df['weight_kg'] = df[weight_col].apply(convert_to_kg)

df.head()
```

```python
# Step 5
region_map = {
    'e cape': 'Eastern Cape',
    'eastern cape': 'Eastern Cape',
    'western cape': 'Western Cape',
    'kzn': 'KwaZulu Natal',
    'kwazulu natal': 'KwaZulu Natal',
    'gauteng': 'Gauteng'
}
df['region'] = df['region'].str.strip().str.lower().map(region_map)

df['client_name'] = df['client_name'].str.strip().str.title()

df.head()
```

```python
# Step 6
df['weight_kg'] = pd.to_numeric(df['weight_kg'], errors='coerce')
df['cost_zar'] = pd.to_numeric(df['cost_zar'], errors='coerce')

df.head()
```

```python
# Step 7
text_cols = ['waste_type', 'driver_name', 'disposal_site', 'remarks']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()

    df.head()
```

```python
# Step 8
df['cost_per_kg'] = df['cost_zar'] / df['weight_kg']

df.head()
```

```python
# Step 9
df.to_csv("waste_collection_cleaned.csv", index=False)

print("\n Data cleaning complete! File saved as 'waste_collection_cleaned.csv'.")
display(df.head(10))
```
